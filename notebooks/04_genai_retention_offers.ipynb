```python
%sql
USE CATALOG ai_job_lakehouse;
USE SCHEMA gold;

-- Generate AI-powered retention offers for the top 5 highest-risk active customers
CREATE OR REPLACE TABLE genai_retention_action_plan AS
SELECT
    customer_id,
    Contract AS contract_type,
    InternetService AS internet_service,
    tenure AS tenure_months,
    MonthlyCharges AS monthly_charges,
    churn_probability,
    ai_risk_segment,
    ai_query(
        'databricks-meta-llama-3-1-8b-instruct',
        CONCAT(
            'You are a telecom retention specialist. Write a concise 2-sentence retention offer for a customer with: ',
            'Contract=', Contract, ', ',
            'Internet=', InternetService, ', ',
            'Tenure=', tenure, ' months, ',
            'MonthlyCharges=$', MonthlyCharges, ', ',
            'ChurnRisk=', churn_probability, '%. ',
            'Include a specific discount or plan upgrade incentive.'
        )
    ) AS ai_recommended_offer
FROM ai_job_lakehouse.gold.ml_churn_predictions
WHERE Churn = 'No'
ORDER BY churn_probability DESC
LIMIT 5;

SELECT * FROM genai_retention_action_plan;
```
